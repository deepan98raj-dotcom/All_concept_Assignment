###**Define**

np.dot() → calculate relationships between vectors

Euclidean distance → find users who are close/similar

Cosine similarity → measure similarity between users/movies

np.argsort() → rank the most similar users/movies

````python
# dot

A = np.array([[1,3],[2,4]])

B = np.array([[5,2],[1,2]])

result=np.dot(A,B)
print(result)

# Euclidean distance
user1 = np.array([5, 0, 3, 4, 0])
user2 = np.array([1, 2, 3, 3, 5])

distance = np.linalg.norm(user1 - user2)
print(distance)

# Cosine similarity
use1 = np.array([5, 4, 3])
use2 = np.array([4, 2, 3])

cosine_similarity = np.dot(use1, use2) / (np.linalg.norm(use1) * np.linalg.norm(use2))

print(cosine_similarity)

# argSorting
arr = np.array([3,2,1])
sorted_indices = np.argsort(arr)
print(sorted_indices)

````

````python
import numpy as np

# Movie names
movies = np.array([
    "Avatar",
    "Titanic",
    "Inception",
    "Avengers",
    "Interstellar",
    "Joker"
])

ratings = np.array([
    [5, 4, 0, 5, 0, 3],  # User 0
    [5, 0, 4, 5, 3, 0],  # User 1
    [1, 2, 5, 0, 4, 0],  # User 2
    [4, 5, 0, 4, 0, 3]   # User 3
])


def cosine_similarity(a, b):
    denominator = np.linalg.norm(a) * np.linalg.norm(b)

    if denominator == 0:
        return 0

    return np.dot(a, b) / denominator


# Input user
user_id = int(input("Enter User ID: "))

target_user = ratings[user_id]

# Calculate similarity
similarities = []

for i in range(len(ratings)):

    if i == user_id:
        continue

    similarity = cosine_similarity(
        target_user,
        ratings[i]
    )

    similarities.append((i, similarity))


# Sort users by similarity
similarities.sort(
    key=lambda x: x[1],
    reverse=True
)

print("\nSimilar users:")
for user, score in similarities:
    print("User", user, "Similarity:", round(score, 2))


# Calculate recommendation scores
recommendation_scores = np.zeros(len(movies))

for similar_user, similarity in similarities:

    for movie_index in range(len(movies)):

        # Only consider movies not watched
        if target_user[movie_index] == 0:

            recommendation_scores[movie_index] += (
                similarity *
                ratings[similar_user][movie_index]
            )


# Sort movie indices by recommendation score
ranked_movies = np.argsort(
    recommendation_scores
)[::-1]


# Get top 3 unwatched movies
recommendations = []

for movie_index in ranked_movies:

    if target_user[movie_index] == 0:

        recommendations.append(movie_index)

    if len(recommendations) == 3:
        break


print("\nTop 3 Recommended Movies:")

for movie_index in recommendations:

    print(
        movies[movie_index],
        "Score:",
        round(recommendation_scores[movie_index], 2)
    )
````

#  Student Marks Analysis & Gradebook
Build an automated grading script using basic classroom data. You can create a mock dataset using Python dictionaries and convert it directly into a DataFrame.


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

students = {
    'student':['arun','sai','priya','rahul','sneha'],
    'math':[85,72,67,89,92],
    'science':[90,78,88,75,82],
    'history':[95,75,92,65,70]
}

df = pd.DataFrame(students)
print(df.head())
df['Total'] = df[['math','science','history']].sum(axis=1)

df['Average'] = df[['math','science','history']].mean(axis=1)

def cal_grade(avg):
  if avg >= 90:
    return 'A+'
  elif avg >= 80:
    return 'A'
  elif avg >= 70:
    return 'B'
  elif avg >= 60:
    return 'C'
  elif avg >= 50:
    return 'D'
  else:
    return 'F'

df['Grade'] = df['Average'].apply(cal_grade)
df['Average'] = df['Average'].round(2)

df = df.sort_values('Average',ascending=False)

print('STUDENT GRADEBOOK')
print(df)

print('\n Top Student:',df.iloc[0]['student'])

print('\n Students with A grade or above:')
print(df[df['Grade'] >= '80'])
````
